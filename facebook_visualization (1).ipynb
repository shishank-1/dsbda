```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset
df = pd.read_csv('fb.csv')


df.head()
```

```python
# Dataset Info
df.info()

```

```python
df.describe()
```

```python
# Post Type Count
sns.countplot(x='Type', data=df)
plt.title("Count of Facebook Post Types")
#plt.xticks(rotation=45)
plt.xticks()
plt.show()
```

```python
# Total Interactions Distribution
plt.hist(df['Total Interactions'], bins=15)
plt.title("Distribution of Total Interactions")
plt.xlabel("Total Interactions")
plt.ylabel("Frequency")
plt.show()
```

```python
# Paid vs Unpaid Posts Interaction
sns.boxplot(x='Paid', y='Total Interactions', data=df)
plt.title("Paid vs Unpaid Posts Total Interactions")
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(14,8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
# Post Month vs Total Reach
sns.barplot(x='Post Month', y='Lifetime Post Total Reach', data=df)
plt.title("Post Month vs Lifetime Post Total Reach")
plt.show()
```

```python
# Likes vs Shares Scatter Plot
plt.scatter(df['like'], df['share'])
plt.xlabel("Likes")
plt.ylabel("Shares")
plt.title("Likes vs Shares")
plt.show()
```

```python
# Type vs Average Engagement Users
sns.barplot(x='Type', y='Lifetime Engaged Users', data=df)
plt.title("Post Type vs Average Lifetime Engaged Users")
plt.xticks(rotation=45)
plt.show()
```
