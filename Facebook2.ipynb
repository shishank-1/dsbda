# Facebook Dataset Visualization Analysis

Visualization notebook for Facebook dataset using Matplotlib and Seaborn.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load Dataset
df = pd.read_csv('fb.csv')

# Show First 5 Rows
df.head()
```

```
# Dataset Information
df.info()
```

## Line Plot

```
plt.figure(figsize=(8,5))

plt.plot(
    df['Page total likes'],
    df['Category'],
    color='royalblue',
    linewidth=2
)

plt.xlabel('Page total likes')
plt.ylabel('Category')
plt.title('Page total likes vs Category')

plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## Bar Plot

```
plt.figure(figsize=(8,5))

df['Type'].value_counts().plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.xlabel('Type')
plt.ylabel('Count')
plt.title('Type Distribution')

plt.show()
```

## Histogram

```
plt.figure(figsize=(8,5))

plt.hist(
    df['Page total likes'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.xlabel('Page total likes')
plt.ylabel('Frequency')
plt.title('Page total likes Distribution')

plt.show()
```

## Scatter Plot with Regression Line

```
x = df['Page total likes']
y = df['Category']

plt.figure(figsize=(8,5))

plt.scatter(
    x,
    y,
    color='mediumseagreen',
    edgecolor='black',
    alpha=0.7
)

# Regression Line
m, b = np.polyfit(x, y, 1)

plt.plot(
    x,
    m*x + b,
    color='red',
    linewidth=2
)

plt.xlabel('Page total likes')
plt.ylabel('Category')
plt.title('Page total likes vs Category')

plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## Pie Chart

```
data = df['Type'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    data,
    labels=data.index,
    autopct='%1.1f%%'
)

plt.title('Type Distribution')

plt.show()
```

## Box Plot

```
plt.figure(figsize=(6,5))

plt.boxplot(
    df['Category'],
    patch_artist=True,
    boxprops=dict(facecolor='lightblue')
)

plt.ylabel('Category')
plt.title('Category Box Plot')

plt.show()
```

## Correlation Heatmap

```
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap')

plt.show()
```

## Count Plot

```
plt.figure(figsize=(8,5))

sns.countplot(x='Type', data=df)

plt.title('Type Count Plot')

plt.show()
```
