````python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://books.toscrape.com/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

products = soup.find_all("article", class_="product_pod")

data = []

for product in products:

    # Product Name
    name = product.h3.a["title"]

    # Product Link
    product_link = "https://books.toscrape.com/" + product.h3.a["href"]

    # Image Link
    image = "https://books.toscrape.com/" + product.img["src"]

    # Rating
    rating = product.find("p")["class"][1]

    # Price
    price = product.find("p", class_="price_color").text

    # Availability
    availability = product.find(
        "p",
        class_="instock availability"
    ).text.strip()

    data.append({
        "Name": name,
        "Price": price,
        "Rating": rating,
        "Availability": availability,
        "Product Link": product_link,
        "Image Link": image
    })

df = pd.DataFrame(data)

print(df.head())
````

````python
IMPORTANT CONCEPTS FOR E-COMMERCE REVIEW SCRAPER (WITH ANSWERS)

1. What is Web Scraping?

Web scraping is the process of extracting data from websites automatically using programs or scripts.

Example:
Fetching product names, prices, ratings, and reviews from an e-commerce website.

---

2. Why is Web Scraping Used?

Web scraping is used for:

* Price comparison
* Sentiment analysis
* Market research
* Product recommendation systems
* Review analysis

---

3. What is BeautifulSoup?

BeautifulSoup is a Python library used to parse HTML and XML documents.

It helps:

* Find HTML tags
* Extract text
* Navigate webpage structure

Example:
soup.find("p")

---

4. What is Requests Library?

Requests is a Python library used to send HTTP requests to websites.

Example:
response = requests.get(url)

---

5. What is HTML Parsing?

HTML parsing means analyzing webpage source code and extracting useful information from tags.

Example tags:

<h1>
<p>
<div>
<a>

---

6. What is a Tag in HTML?

A tag defines webpage elements.

Example:

<p>Price</p>

Here, p is a paragraph tag.

---

7. Difference Between find() and find_all()

find()

* Finds first matching element

find_all()

* Finds all matching elements

Example:
soup.find("p")

soup.find_all("p")

---

8. What is a Class in HTML?

Class is used to identify HTML elements.

Example:

<p class="price_color">

Access in Python:
soup.find("p", class_="price_color")

---

9. What is CSS Selector?

CSS selectors are patterns used to select HTML elements.

Example:
soup.select(".price_color")

---

10. What is HTTP Request?

HTTP request is communication sent from client to server.

Types:

* GET
* POST

---

11. What is GET Request?

GET request fetches data from server.

Example:
requests.get(url)

---

12. What is User-Agent?

User-Agent tells the website which browser/device is sending request.

Used to avoid blocking.

Example:
headers = {
"User-Agent": "Mozilla/5.0"
}

---

13. Why Use Headers in Scraping?

Headers help:

* Mimic real browser
* Avoid anti-bot detection
* Improve request success rate

---

14. What is Data Cleaning?

Data cleaning removes:

* Null values
* Duplicate data
* Incorrect formatting

Example:
df.drop_duplicates()

---

15. What is Pandas?

Pandas is a Python library used for:

* Data analysis
* DataFrames
* CSV handling

---

16. What is DataFrame?

DataFrame is a table-like structure in pandas.

Example:
| Name | Price |

---

17. How to Save Scraped Data?

Using CSV file.

Example:
df.to_csv("products.csv")

---

18. What is Real-Time Scraping?

Fetching latest live data directly from website during execution.

---

19. What is Dynamic Website?

Dynamic websites load data using JavaScript.

Normal requests may fail.

Need:

* Selenium
* Playwright

---

20. What is Selenium?

Selenium automates web browsers.

Used for:

* Dynamic websites
* Clicking buttons
* Login automation

---

21. Difference Between BeautifulSoup and Selenium

BeautifulSoup:

* Parses HTML
* Faster
* Best for static pages

Selenium:

* Controls browser
* Slower
* Best for dynamic pages

---

22. What is CAPTCHA?

CAPTCHA prevents bots from scraping websites.

Example:
“I am not a robot”

---

23. Why Some Websites Block Scraping?

Reasons:

* Security
* Server load
* Data protection

---

24. What is robots.txt?

robots.txt tells bots:

* Allowed pages
* Restricted pages

---

25. Ethical Issues in Web Scraping

Must avoid:

* Overloading servers
* Scraping private data
* Violating website terms

---

26. What is Rate Limiting?

Limiting number of requests.

Used to avoid blocking.

Example:
time.sleep(2)

---

27. What is Sentiment Analysis?

Analyzing reviews/comments to detect:

* Positive
* Negative
* Neutral sentiment

---

28. What is Comment Tagging?

Assigning labels like:

* Useful
* Recommended
* Bad Quality

---

29. What are Common Challenges in Scraping?

* CAPTCHA
* Dynamic content
* IP blocking
* HTML structure changes

---

30. Advantages of Web Scraping

* Automation
* Fast data collection
* Real-time monitoring
* Large-scale analysis

---

31. Limitations of Web Scraping

* Website blocking
* Legal restrictions
* Dynamic content handling
* Maintenance required

---

32. What is JSON?

JSON stores data in key-value format.

Example:
{
"name": "Laptop",
"price": 50000
}

---

33. What is API?

API provides structured data directly from server.

Better than scraping when available.

---

34. What is the Workflow of Your Project?

Steps:

1. Send request
2. Fetch HTML
3. Parse webpage
4. Extract product data
5. Clean data
6. Store in DataFrame
7. Export CSV

---

35. Why did you choose BooksToScrape?

“BooksToScrape was chosen because it is designed for educational scraping practice and does not contain anti-scraping restrictions.”
````
